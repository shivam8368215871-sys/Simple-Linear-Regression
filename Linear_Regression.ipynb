Simple linear regression is a statistical and machine learning method used to model the relationship between a single independent variable (X) and a dependent target variable (Y) by fitting a straight line through the data

```python
import numpy as np
import matplotlib.pyplot as plt 
import pandas as pd
```

```python
df = pd.read_csv('/home/shivam/Documents/vs code insider/data/placement.csv')
df.head()
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.title("CGPA and Package graph")
plt.xlabel("CGPA")
plt.ylabel("Package")
plt.show()
```

```python
x = df.iloc[:,0:1]
y = df.iloc[:,-1]
print(x)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train,y_train)
```

```python
lr.predict(x_test.iloc[1].values.reshape(1,1))
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.plot(x_test,lr.predict(x_test),color='red')
plt.title("CGPA and Package graph")
plt.xlabel("CGPA")
plt.ylabel("Package")
plt.show()
```
